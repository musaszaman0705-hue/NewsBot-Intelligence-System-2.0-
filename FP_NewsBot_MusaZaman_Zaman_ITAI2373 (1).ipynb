```python
class SentimentEvolutionTracker:

    def __init__(self):

        self.model = pipeline(
            "sentiment-analysis"
        )


    def analyze_sentiment(self, article_text):

        result = self.model(article_text[:512])[0]

        return {
            "sentiment": result["label"],
            "confidence": round(result["score"],3)
        }
```

```python
class TopicDiscoveryEngine:

    def __init__(self):

        self.model = BERTopic()


    def fit_topics(self, documents):

        topics, probs = self.model.fit_transform(
            documents
        )

        return topics


    def get_topics(self):

        return self.model.get_topic_info()
```

```python
class EntityRelationshipMapper:

    def __init__(self):

        self.nlp = spacy.load(
            "en_core_web_sm"
        )


    def extract_entities(self,text):

        doc=self.nlp(text)

        entities=[]

        for ent in doc.ents:

            entities.append(
                {
                "text":ent.text,
                "type":ent.label_
                }
            )

        return entities
```

```python
class SemanticSearchEngine:


    def __init__(self):

        self.encoder=SentenceTransformer(
            "all-MiniLM-L6-v2"
        )


    def encode_documents(self,documents):

        return self.encoder.encode(
            documents
        )


    def search(self,query,documents):

        doc_vectors=self.encode_documents(
            documents
        )

        query_vector=self.encoder.encode(
            [query]
        )

        scores=cosine_similarity(
            query_vector,
            doc_vectors
        )[0]


        index=np.argmax(scores)

        return documents[index]
```

```python
from transformers import pipeline

class NewsBot2:

    def __init__(self):
        self.name = "NewsBot 2.0"
        # Initialize pipelines as instance attributes
        self.summarizer = pipeline(
            "summarization",
            model="facebook/bart-large-cnn"
        )
        self.sentiment_analyzer = pipeline(
            "sentiment-analysis",
            model="distilbert-base-uncased-finetuned-sst-2-english"
        )
        self.qa_model = pipeline(
            "question-answering",
            model="distilbert-base-cased-distilled-squad"
        )


    def analyze(self, article):

        # Use instance attributes
        summary = self.summarizer(
            article,
            max_length=100,
            min_length=30,
            do_sample=False
        )


        # Use instance attributes
        sentiment = self.sentiment_analyzer(article)


        return {
            "summary": summary[0]["summary_text"],
            "sentiment": sentiment[0]["label"],
            "confidence": round(sentiment[0]["score"],3)
        }


    def chat(self, question, article):

        # Use instance attributes
        answer = self.qa_model(
            question=question,
            context=article
        )

        return answer["answer"]
```

```python
INTENTS = {

    "summary": [
        "summarize",
        "summary",
        "short version"
    ],

    "sentiment":[
        "sentiment",
        "positive",
        "negative"
    ],

    "entities":[
        "who",
        "organization",
        "people",
        "companies"
    ],

    "translate":[
        "translate"
    ],

    "search":[
        "find",
        "search",
        "show"
    ],

    "topics":[
        "topic",
        "topics",
        "trend"
    ]
}
```

```python
class IntentClassifier:

    def classify(self, query):

        q = query.lower()

        for intent, words in INTENTS.items():

            for word in words:

                if word in q:

                    return intent

        return "general"
```
